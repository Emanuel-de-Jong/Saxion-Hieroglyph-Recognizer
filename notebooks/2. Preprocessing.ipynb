## Imports

```python
import os
import pickle

from sklearn.preprocessing import LabelEncoder, MinMaxScaler, StandardScaler
```

## Load input

```python
with open(f'..{os.sep}dataset-numpy{os.sep}features_after_extraction.pkl', 'rb') as f:
    features = pickle.load(f)

features.iloc[:5]
```

## Global

#### Variables

```python
label_columns = ['label', 'label_num']
```

#### Functions

```python
def get_features_without_labels():
    return features.drop(columns=label_columns)

def get_feature_columns():
    return get_features_without_labels().columns.tolist()
```

## Make numerical version of labels
Some algorithms need it

```python
features['label_num'] = LabelEncoder().fit_transform(features['label'])
```

## Scale numeric features

Between the features, values that contain clear outliers are scaled with a z-score scaler while values without clear outliers are scaled via the MinMax scaler. This decision is based on the information found when using describe() on the dataset and reviewing the min and max values compared to the 75%, 50% and 25%.

```python
features_to_min_max = ['height', 'width', 'centroid_x', 'centroid_y', 'extent', 'perimeter']
features_to_z_score = ['area', 'hole_count', 'convex_area', 'solidity', 'eccentricity', 'orientation',
                       'hu_1', 'hu_2', 'hu_3', 'hu_4', 'hu_5', 'hu_6', 'hu_7', 'top_bottom_ratio', 'height_width_ratio']

min_max_scaler = MinMaxScaler()
z_score_scaler = StandardScaler()

print(features[features_to_z_score[0]][:5])

if len(features_to_min_max) > 0:
    to_scale = features[features_to_min_max]
    scaled_values = min_max_scaler.fit_transform(to_scale)
    features[features_to_min_max] = scaled_values

    print(features[features_to_min_max[0]][:5])

if len(features_to_z_score) > 0:
    to_scale = features[features_to_z_score]
    scaled_values = z_score_scaler.fit_transform(to_scale)
    features[features_to_z_score] = scaled_values

    print(features[features_to_z_score[0]][:5])
```

#### Save scalers
For use in the webserver.

```python
with open(f'..{os.sep}webserver{os.sep}models{os.sep}min_max_scaler.pkl', 'wb') as f:
    pickle.dump(min_max_scaler, f)

with open(f'..{os.sep}webserver{os.sep}models{os.sep}z_score_scaler.pkl', 'wb') as f:
    pickle.dump(z_score_scaler, f)
```
